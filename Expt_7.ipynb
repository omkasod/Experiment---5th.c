Case Study 2: Accessing Dictionary Items

```
d={"name":"MSD",1:"Python",(1,2):[1,2,4]}
# Access using key
print(d["name"])
# Access using get()
print(d.get("name"))
```

Case Study 3: Adding and Updating Dictionary Items

```
d={1:'Zeal',2:'College of',3:'Engineering'}

# Adding a new key-value pair
d["Established in"]=1980

# Updating an existing value
d[1]="Imperial"
print(d)
```

Case Study 4: Removing Dictionary Items

```
d={1:'Zeal',2:'Engineering',3:'College','Established in':2008}
print(d)

# Using del
del d["Established in"]
print(d)

# Using pop()
val=d.pop(1)
print(val)

# Using popitem()
key,val=d.popitem()
print(f"Key: {key}, Value: {val}")

# Using clear()
d.clear()
print(d)
```

Case Study 5: Iterating Through a Dictionary

```
d={1:'Zeal',2:'Engineering','Established in':2008}

# Iterate over keys
for key in d:
  print(key)

# Iterate over values
for value in d.values():
  print(value)

# Iterate over key-value pairs
for key,value in d.items():
  print(f"{key}: {value}")
```

Case Study 7: Student Record System

```
students={
101:{"name":"Mahi","marks":85},
102:{"name":"Virat","marks":90},
103:{"name":"Rohit","marks":78}
}

# Access student details
print(students[101]["name"])

# Add new student
students[104]={"name":"Priya","marks":88}

# Update marks
students[102]["marks"]=95

# Display all students
for roll,details in students.items():
  print(roll,details)
```

Case Study 8: Word Frequency Counter

```
text = "apple banana apple mango banana apple"
words = text.split()
frequency = {}
for word in words:
  if word in frequency:
    frequency[word] += 1
  else:
    frequency[word] = 1
print(frequency)
```
